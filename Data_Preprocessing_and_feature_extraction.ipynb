```python
import os
import pandas as pd
from tsfresh.feature_extraction import extract_features

# Path to the main folder containing all subfolders
main_folder = "/content/drive/MyDrive/Raeyaan shared folder/Exoskeleton Project/Dataset/Raeyaan Dataset" # Change this to your actual folder path

# List to store processed data
final_data = []

# Iterate through each folder in the main directory
for folder in os.listdir(main_folder):
    folder_path = os.path.join(main_folder, folder)

    if os.path.isdir(folder_path):  # Ensure it's a directory
        for file in os.listdir(folder_path):
            if file.endswith(".csv"):  # Process only CSV files
                file_path = os.path.join(folder_path, file)

                # Read the CSV file
                df = pd.read_csv(file_path)

                # ✅ Format CSV: Add 'id' and 'time' columns
                df["id"] = range(1, len(df) + 1)  # 1 to 60
                df["time"] = range(1, len(df) + 1)  # 1 to 60

                # ✅ Keep only 'id', 'time', and 'emg_envelop' columns
                if 'exo_project_emg_1_emg_envelop' not in df.columns:
                    print(f"Skipping {file} (no 'exo_project_emg_1_emg_envelop' column)")
                    continue

                df = df[["id", "time", "exo_project_emg_1_emg_envelop"]]

                # ✅ Extract features using tsfresh (only on 'emg_envelop')
                extracted_features = extract_features(df, column_id='id', column_sort='time')

                # Flatten extracted features into a single row
                row_data = extracted_features.iloc[0].tolist()

                # Append folder name as a new column
                row_data.append(folder)

                # Store the processed row
                final_data.append(row_data)

# ✅ Create a DataFrame with extracted features
feature_columns = list(extracted_features.columns)  # Get feature names
feature_columns.append("Folder")  # Add folder column

final_df = pd.DataFrame(final_data, columns=feature_columns)

# ✅ Save the final DataFrame as a CSV file
final_csv_path = os.path.join(main_folder, "tsfresh_updated.csv")
final_df.to_csv(final_csv_path, index=False)

print(f"Feature CSV saved at: {final_csv_path}")
```

# Method 3: Featurization using Librosa and extracting MFCC features

```python
!pip install librosa -q
```

```python
import os
import pandas as pd
import numpy as np
import librosa

# Path to the main folder containing all subfolders
main_folder = "/content/drive/MyDrive/Raeyaan shared folder/Exoskeleton Project/Dataset/Raeyaan Dataset"  # Change this to your actual folder path

# List to store processed data
final_data = []

# Number of MFCC coefficients to extract
n_mfcc = 20  # You can adjust this based on your needs

# Iterate through each folder in the main directory
for folder in os.listdir(main_folder):
    folder_path = os.path.join(main_folder, folder)

    if os.path.isdir(folder_path):  # Ensure it's a directory
        for file in os.listdir(folder_path):
            if file.endswith(".csv"):  # Process only CSV files
                file_path = os.path.join(folder_path, file)

                # Read the CSV file
                df = pd.read_csv(file_path)

                # ✅ Ensure the 'emg_envelop' column exists
                if 'exo_project_emg_1_emg_envelop' not in df.columns:
                    print(f"Skipping {file} (no 'exo_project_emg_1_emg_envelop' column)")
                    continue

                # Convert EMG data to numpy array
                emg_signal = df["exo_project_emg_1_emg_envelop"].values.astype(float)

                # Compute MFCC features (treating EMG as a signal)
                mfccs = librosa.feature.mfcc(y=emg_signal, sr=1000, n_mfcc=n_mfcc)  # Assuming sr=1000 Hz

                # Compute statistical features for each MFCC coefficient
                mfcc_features = np.mean(mfccs, axis=1)  # Taking mean of each coefficient

                # Append folder name as a new column
                row_data = mfcc_features.tolist()
                row_data.append(folder)

                # Store the processed row
                final_data.append(row_data)

# ✅ Create a DataFrame with extracted MFCC features
mfcc_columns = [f"MFCC_{i+1}" for i in range(n_mfcc)]  # Create column names
mfcc_columns.append("Folder")  # Add folder column

final_df = pd.DataFrame(final_data, columns=mfcc_columns)

# ✅ Save the final DataFrame as a CSV file
final_csv_path = os.path.join(main_folder, "mfcc.csv")
final_df.to_csv(final_csv_path, index=False)

print(f"MFCC feature CSV saved at: {final_csv_path}")
```
